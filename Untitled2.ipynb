```python
import seaborn as sns
titanic = sns.load_dataset("titanic")
print(titanic.head())
# نسبة النجاة حسب الجنس
نسبة_النجاة = titanic.groupby("sex")["survived"].mean()
print(نسبة_النجاة)
نسبة_النجاة_حسب_الدرجة = titanic.groupby("pclass")["survived"].mean()
print(نسبة_النجاة_حسب_الدرجة)
import matplotlib.pyplot as plt

نسبة_النجاة_حسب_الدرجة.plot(kind="bar")
plt.xlabel("الدرجة")
plt.ylabel("نسبة النجاة")
plt.title("نسبة النجاة حسب الدرجة")
plt.show()
```

```python
diamonds = sns.load_dataset("diamonds")
print(diamonds.head())
متوسط_السعر_حسب_الجودة = diamonds.groupby("cut")["price"].mean()
print(متوسط_السعر_حسب_الجودة)
متوسط_السعر_حسب_الجودة.plot(kind="bar")
plt.xlabel("جودة القص")
plt.ylabel("متوسط السعر")
plt.title("متوسط سعر الألماس حسب جودة القص")
plt.show()
متوسط_الوزن_حسب_الجودة = diamonds.groupby("cut")["carat"].mean()
print(متوسط_الوزن_حسب_الجودة)
```
